```python
import pandas as pd
import numpy as np

print("--- Step 1: Read and Add Column ---")
# 1. Load cleaned_sales.csv with pd.read_csv()
df = pd.read_csv('/content/cleaned_sales.csv')

# Add tax_amount equal to 5% of revenue
df['tax_amount'] = df['revenue'] * 0.05
print("First 5 rows of loaded dataset:")
display(df.head())

print("\n--- Step 2: Update and Remove ---")
# Use .loc[] to increase unit_price of Books items by 10%
# First, check if 'category' column exists and look for 'Books' or 'books'
category_col = [col for col in df.columns if col.lower() == 'category'][0]
price_col = [col for col in df.columns if 'price' in col.lower()][0]

# Normalize checking for category 'Books' case-insensitively
books_mask = df[category_col].str.lower() == 'books'
print(f"Number of Books items before price update: {books_mask.sum()}")

# Increase unit_price of Books items by 10%
df.loc[books_mask, price_col] = df.loc[books_mask, price_col] * 1.10

# Recalculate revenue (quantity * unit_price) and tax_amount
qty_col = [col for col in df.columns if 'qty' in col.lower() or 'quantity' in col.lower()][0]
df['revenue'] = df[qty_col] * df[price_col]
df['tax_amount'] = df['revenue'] * 0.05

# Remove one selected row using an existing transaction_id
trans_col = [col for col in df.columns if 'transaction' in col.lower() or 'id' in col.lower()][0]
target_id = df[trans_col].iloc[0]
print(f"Removing row with {trans_col}: {target_id}")
df = df[df[trans_col] != target_id]

# Create a temporary review_status column, then remove it
df['review_status'] = 'Pending'
print("Temporary column 'review_status' added. Columns:", list(df.columns))
df = df.drop(columns=['review_status'])
print("Temporary column 'review_status' dropped. Columns:", list(df.columns))

print("\n--- Step 3: Save and Verify ---")
# Save to sales_updated.csv
df.to_csv('/content/sales_updated.csv', index=False)

# Read it again, and show the first five rows and its final column names
final_df = pd.read_csv('/content/sales_updated.csv')
print("Final Columns:", list(final_df.columns))
print("First 5 rows of final saved file:")
display(final_df.head())
```
